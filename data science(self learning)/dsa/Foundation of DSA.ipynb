# digits put in the list

```python
n = 43987
li = []
while n>0:
    last = n%10
    li.append(last)
    n = n//10
print(li)
```

```python
num = 43987
digits = []

while num > 0:
    digit = num % 10      # last digit
    digits.append(digit)
    num = num // 10       # remove last digit

digits.reverse()           # correct order
print(digits)
```
