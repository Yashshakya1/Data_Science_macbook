# Time and Space complexity cheat sheet 
* Time Complexity
| Pattern                | Complexity     |
| ---------------------- | -------------- |
| Simple loop            | **O(n)**       |
| Nested loop            | **O(n²)**      |
| Divide by 2 or 10 loop | **O(log n)**   |
| Loop + log loop        | **O(n log n)** |
| Two independent loops  | **O(n + m)**   |
| Constant operations    | **O(1)**       |
| Triple nested loops    | **O(n³)**      |


# SPACE COMPLEXITY CHEAT SHEET
| Pattern               | Complexity |
| --------------------- | ---------- |
| Only variables        | **O(1)**   |
| List of size n        | **O(n)**   |
| 2D array (n×n)        | **O(n²)**  |
| Recursion depth n     | **O(n)**   |
| Hash map with n items | **O(n)**   |

# digits put in the list

| Complexity Type      | Value        |
| -------------------- | ------------ |
| **Time Complexity**  | **O(log n)** |
| **Space Complexity** | **O(log n)** |


```python
n = 43987
li = []
while n>0:
    last = n%10
    li.append(last)
    n = n//10
print(li)
```

| Complexity Type      | Value        |
| -------------------- | ------------ |
| **Time Complexity**  | **O(log n)** |
| **Space Complexity** | **O(log n)** |


```python
num = 43987
digits = []

while num > 0:
    digit = num % 10      # last digit
    digits.append(digit)
    num = num // 10       # remove last digit

digits.reverse()           # correct order
print(digits)
```
