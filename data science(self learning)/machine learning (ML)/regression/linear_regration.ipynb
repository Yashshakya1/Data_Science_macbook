# 🎯 What is Regression?

* Regression is a Supervised Machine Learning technique used to predict continuous values (numbers).

# In simple terms:

* Regression helps you find relationships between variables and predict numeric outcomes.

| Study Hours | Exam Score |
| ----------- | ---------- |
| 1           | 40         |
| 2           | 50         |
| 3           | 60         |
| 4           | 70         |
| 5           | 80         |

# real life example
| Problem             | Input (X)                | Output (Y)       |
| ------------------- | ------------------------ | ---------------- |
| Predict house price | Area, bedrooms, location | Price (₹)        |
| Predict salary      | Experience, education    | Salary (₹)       |
| Predict temperature | Time, location           | Temperature (°C) |
| Predict sales       | Marketing spend          | Sales amount     |


# regression formula 
* y = mx + c 

# types of linear regression 

| Type                           | Description                       | Output            |
| ------------------------------ | --------------------------------- | ----------------- |
| **Linear Regression**          | Predicts using a straight line    | Continuous value  |
| **Multiple Linear Regression** | Uses more than one feature        | Continuous value  |
| **Polynomial Regression**      | Uses curved (non-linear) relation | Continuous value  |
| **Ridge/Lasso Regression**     | Regularized to avoid overfitting  | Continuous value  |
| **Logistic Regression**        | Used for classification (Yes/No)  | Probability (0–1) |


# 🔥 Difference Between array() and asarray()
* ✔ np.array()
* Always creates a new copy of the data (unless told otherwise).
* ✔ np.asarray()
* Creates a view, not a copy, whenever possible → faster.

```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
# data
x = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([10,30,50,70,90])

# model
model = LinearRegression()
model.fit(x,y)
# Prediction
y_pred = model.predict(x)
# plotting
plt.scatter(x,y, color = 'red', label = 'Actual data')
plt.plot(x,y_pred ,color = 'green' ,label ='fit line')
plt.title('example of linear regression')
plt.xlabel('Study hours')
plt.ylabel('Marks')
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Training data
x_axis = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y_axis = np.array([10, 30, 50, 70, 90])

# Create model and fit
model = LinearRegression()
model.fit(x_axis, y_axis)

# Take user input
hours = float(input('Enter your study hours: '))

# Predict
pred = model.predict([[hours]])

# Output
print(f'Study hours: {hours} → Predicted score: {pred[0]:.2f}')
```

```python
q1 = ds.quantile(0.25)
q3 = ds.quantile(0.75)
IQR = q3 - q1 
lower_bound = q1 - 1.5 * IQR 
upper_bound = q3 + 1.5 * IQR

print(IQR)
print(lower_bound)
print(upper_bound)
```

# Feature salection

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
x = pd.DataFrame(ss.fit_transform(x) , columns = x.columns)
```

```python
x
```

# split the data on the basis off test and train 

```python
from sklearn.model_selection import train_test_split 
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2, random_state=2)
print('this shape is before the splitting the data :-',x.shape)
print('After splitting the data :- ',x_train.shape)
print(x_train)
```

# creatign the LinearRegression Model 

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
# -------------------------------------------------------
# 9. PREDICTION SYSTEM
# -------------------------------------------------------

def predict_speed(name, type_1, type_2, hp, attack, defense, sp_attack, sp_defense, weight_kg, total):

    row = [[name, type_1, type_2, hp, attack, defense, sp_attack, sp_defense, weight_kg, total]]

    input_df = pd.DataFrame(row, columns=x.columns)

    input_scaled = ss.transform(input_df)

    prediction = lr.predict(input_scaled)

    return prediction[0]




print("\n------------------- PREDICTION -------------------")


pred = predict_speed(
    name=10,
    type_1=3,
    type_2=5,
    hp=80,
    attack=95,
    defense=85,
    sp_attack=100,
    sp_defense=90,
    weight_kg=72,
    total=540
)

print(pred)
```

# find the m and c value 

```python
print('this is the value of C is :-',lr.intercept_)
```

```python
print('This is the value of M')
for col,coef in zip(x.columns,lr.coef_):
    print(f'{col} :- {coef}')
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
y_pred = lr.predict(x_test)
print('performace on this model :-',r2_score(y_test,y_pred))
print('this is the error number :- ',mean_absolute_error(y_test,y_pred))
```

# assumptin of linear regression model 

```python
import matplotlib.pyplot as plt

plt.scatter(y_test, y_pred)
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.show()
```

# Homoscedasticity (constant variance)

```python
error = y_test - y_pred
plt.scatter(y_pred,error)
plt.axhline(0,color = 'red')
plt.show()
```
