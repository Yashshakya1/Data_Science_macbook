# Mini-Batches? gd

# example on this gd
* 10,000 samples → 100 mini-batches  
* Each mini-batch contains 100 samples  

# ⭐ Why Do We Use Mini-Batches?
Mini-batches give the best of both worlds:
* faster than Batch GD
* more stable than SGD
* Hence mini-batch gradient descent is the standard in deep learning.

# ⭐ Formula
* Let dataset size = m, batch size = b:
* Number of mini-batches per epoch:
* steps = m / b
# example on this formula 
* m = 10,000, b = 100
* steps = 10,000 / 100 = 100 steps
