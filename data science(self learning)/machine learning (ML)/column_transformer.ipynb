# Column Transformer
* “ColumnTransformer is used to handle mixed data types by assigning specific transformations (like scaling, encoding, or imputation) to selected columns, while keeping everything inside one clean preprocessing pipeline.”

| Column    | Type        |
| --------- | ----------- |
| age       | numeric     |
| salary    | numeric     |
| gender    | categorical |
| city      | categorical |
| purchased | target      |
* Har column ko alag preprocessing chahiye.

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
ds = sns.load_dataset('flights')
ds
```

```python
ds['month'].value_counts()
```

```python
x = ds.iloc[:,:-1]
y = ds.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
x_train
```

```python
ds.isnull().sum()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder 
transformer = ColumnTransformer(transformers = [
('t1',OrdinalEncoder(categories = [['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']]),['month'])
],remainder = 'passthrough')
```

```python
print(transformer.fit_transform(x_train).shape)
transformer.fit_transform(x_train)
# transformer.transform(y_train)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```
