# What is hierarchical clustering
* Hierarchical clustering is an unsupervised learning method that builds a hierarchy (tree) of clusters. In the agglomerative approach, each data point starts as its own cluster and the closest clusters are merged step by step until a final cluster structure is formed.

# Linkage Criteria (VERY Important for Interviews)
# These decide how distance between clusters is calculated:
1. Single Linkage – minimum distance between two points across clusters
2. Complete Linkage – maximum distance
3. Average Linkage – average pairwise distance
4. Ward’s Method – merges clusters that increase total variance the least
*Most popular for numerical data

# Dendrogram (Key Concept)
* A dendrogram is a tree-like diagram used to visualize hierarchical clustering.
You cut the dendrogram at a certain height to decide the number of clusters.
# Interview Tip:
# 👉 If they ask “How do you choose the number of clusters?”
* Say: Cut the dendrogram where the vertical distance (height) is maximum.

# 4. Agglomerative vs Divisive Clustering

| Feature    | **Agglomerative**             | **Divisive**          |
| ---------- | ----------------------------- | --------------------- |
| Approach   | Bottom-Up                     | Top-Down              |
| Start      | Each point = separate cluster | One big cluster       |
| Step       | Merge closest clusters        | Split farthest points |
| Popularity | **Most used**                 | Rare                  |
| Complexity | Lower                         | Higher                |
| Dendrogram | Built by merging              | Built by splitting    |

* Most common Q → “Which is more used?”

# ✅ 1. What is hierarchical clustering?
Answer:

* Hierarchical clustering is an unsupervised learning method that builds a hierarchy (tree) of clusters. In the agglomerative approach, each data point starts as its own cluster and the closest clusters are merged step by step until a final cluster structure is formed.

# ✅ 2. Difference between Agglomerative and Divisive clustering
| Agglomerative                | Divisive                   |
| ---------------------------- | -------------------------- |
| Bottom-up approach           | Top-down approach          |
| Start with individual points | Start with one big cluster |
| Merge clusters               | Split clusters             |
| Most commonly used           | Rarely used                |

# 3. What is a dendrogram?
* A dendrogram is a tree-like diagram that shows how clusters are merged (or split) in hierarchical clustering.

# 4. How do you interpret a dendrogram?
1. X-axis → data points
2. Y-axis → distance (or similarity)
3. Height of merge → how far clusters are
* 👉 Larger vertical gaps = better separation

# 5. How to decide number of clusters from dendrogram?
* Cut the dendrogram at the point where the vertical distance between merges is maximum.
This gives the optimal number of clusters.

# 6. Why is Ward linkage preferred?
* Ward linkage minimizes the within-cluster variance, which results in more compact and well-separated clusters.

# 7. What are linkage methods?
* Linkage methods define how distance between clusters is calculated:
1. Single linkage → minimum distance
2. Complete linkage → maximum distance
3. Average linkage → average distance

# 8. Difference between KMeans and hierarchical clustering
| KMeans              | Hierarchical            |
| ------------------- | ----------------------- |
| Need to specify K   | No need initially       |
| Fast                | Slow                    |
| Works on large data | Not good for large data |
| Uses centroids      | Uses distance hierarchy |

# 9. When would you NOT use hierarchical clustering?
* When dataset is very large

# 10. Complexity of Agglomerative clustering
* Time complexity is O(n²) (sometimes O(n³) depending on implementation).
# Why slow?
* Because it computes and updates a distance matrix for all data points repeatedly.
# 11. If dataset has 1M rows, which clustering will you use and why?
* I would use K-Means because it is much faster (O(n)) and scalable, whereas hierarchical clustering is too slow for large datasets.

# Agglomerative Hierarchical Clustering (ML Interview Definition)
* Agglomerative Hierarchical Clustering is a bottom-up clustering algorithm in which each data point starts as its own cluster, and then pairs of clusters are repeatedly merged based on similarity (or minimum distance) until all points form one big cluster or a desired number of clusters is reached.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# Sample dataset (5 points)
X = np.array([
    [1, 2],
    [2, 3],
    [5, 8],
    [6, 9],
    [1, 0]
])

# Perform hierarchical/agglomerative clustering
Z = linkage(X, method='ward')

# Plot the dendrogram
plt.figure(figsize=(8, 5))
dendrogram(Z)
plt.title("Dendrogram Example - Ward Linkage")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.show()
```
