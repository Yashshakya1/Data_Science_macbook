# What is DBSCAN

#  DBSCAN is stand for density based spatial clustring og application with Noise 
* DBSCAN is a density based clustring algorithum that form a group based on density region of points and automatically identifiesthe noise/outliear without any required number of cluster in advance 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.cluster import DBSCAN
```

```python
ds = sns.load_dataset("iris")
x = ds.iloc[:,:-1]
y = ds.iloc[:,-1]
```

```python
dbscan = DBSCAN()
dbscan.fit_predict(x)
```
