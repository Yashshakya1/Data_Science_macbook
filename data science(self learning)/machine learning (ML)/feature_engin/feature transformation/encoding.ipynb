# endcoding 

 * encoding means converting the data from one format into onother format is known as encoding 
 * why need this because ml don't read the string they only know the numbers like 0 or 1
 * types of encoding:-
 * 1. label encoding
 * 2. one hot encodign
 * 3. Ordinal Encoding (Used when the categories have a natural order.) like :-
 * | Education   | Encoded |
| ----------- | ------- |
| High School | 1       |
| Bachelor’s  | 2       |
| Master’s    | 3       |
| PhD         | 4       |

* 4️⃣ Frequency / Count Encoding (Replace each category with how often it appears.)
| City   | Frequency |
| ------ | --------- |
| Delhi  | 3         |
| Mumbai | 2         |
| Delhi  | 3         |
| Pune   | 1         |

* 5️⃣ Binary / Hash / Target Encoding (Used for large datasets or high-cardinality features (e.g., 1000+ categories).)

   # this is summary tabel
* | Encoding Type         | Best for             | Description                          |
| --------------------- | -------------------- | ------------------------------------ |
| **Label Encoding**    | Simple categories    | Assigns numeric labels (0, 1, 2, …)  |
| **One-Hot Encoding**  | Unordered categories | Creates binary columns               |
| **Ordinal Encoding**  | Ordered categories   | Keeps natural rank (low → high)      |
| **Count / Frequency** | Large categories     | Uses counts of occurrences           |
| **Target Encoding**   | Advanced ML          | Uses relationship to target variable |

# ordinary encoding 

```python
import seaborn as sns 
from sklearn.model_selection import train_test_split
print(sns.get_dataset_names())
dataset = sns.load_dataset('tips')
dataset
```

```python
print(dataset['day'].value_counts())
print(dataset['time'].value_counts())
```

```python
x = dataset.iloc[:,0:-1]
y = dataset.iloc[:,-1]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2)
```

```python
x_train
```

```python
from sklearn.preprocessing import OrdinalEncoder
col = ['day','time']
oe = OrdinalEncoder(categories = [['Thur','Fri','Sat','Sun'],['Lunch','Dinner']]) 
# categories parameter is used for arrange the order as you wish
oe.fit_transform(x_train[col])
```

```python
# some attributes 
oe.categories_ # shoing the categories name 
```

# label Encoding 

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

li = ['Red','Blue','Green','Red']
encoding = le.fit_transform(li)
print(encoding)
print(li)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd 
ds = pd.DataFrame({
           'name': ['yash','rani didi','tanisha','suraj','roshni'],
            'gender': ['M','F','F','M','F'],
          'passed': ['yes','yes','no','yes','no']
})
le = LabelEncoder()
ds_label = ds.copy()

ds_label['Gender_endcoded'] = le.fit_transform(ds_label['gender'])
ds_label['Passed_endcoded'] = le.fit_transform(ds_label['passed'])
print(ds_label)
```

# one hot coding
use the encoding there are two ways to execute 
 * pd.get_dummies() is a pandas function used to perform One-Hot Encoding automatically.
 * from sklearn.preprocessing import OneHotEncoder

```python
import seaborn as sns
df = sns.load_dataset('titanic')
df
```

```python
print(df['embarked'].value_counts())
print(df['sex'].value_counts())
```

```python
pd.get_dummies(df,columns=['embarked','sex'])
```

# drop_first=True drop the first columns of each coloumns 

```python
# n-1 encoding 
pd.get_dummies(df,columns=['embarked','sex'],drop_first=True)
```

```python
df.head(2)
```

```python
features = df.iloc[:,:-1]
target = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
train_x,test_x,y_train,y_test = train_test_split(features,target,test_size = 0.2)
train_x.head() 
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first') # drop='first' remove the first category columns for each no of columns
# sparse_output one more paramenter used for showing the array not use the toarray() fuciton
new_x_train = ohe.fit_transform(train_x[['embarked','sex']]).toarray()
# ohe is creating by defulat sparse matrix 
# toarray() this mehod showing the array
print(new_x_train)
```
