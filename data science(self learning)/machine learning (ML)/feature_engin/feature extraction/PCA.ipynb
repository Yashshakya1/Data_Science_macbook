PCA is a dimensionality reduction technique which is used to reduce the number of feature while
keeping the most of the important information

| Problem                        | PCA Solution                   |
| ------------------------------ | ------------------------------ |
| Too many features → model slow | PCA reduces dimensions         |
| Features are correlated        | PCA removes multicollinearity  |
| Hard to visualize              | PCA converts to 2D/3D          |
| Noise in data → accuracy low   | PCA keeps only useful patterns |
