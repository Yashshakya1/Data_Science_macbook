# frequency encoder 

* Frequency Encoding (also called Count Encoding) converts each category into the number of times it appears in the dataset.


* Frequency Encoding replaces each category with the number of times it appears in the dataset. It helps
  * handle high-cardinality features without increasing dimensionality.”

# ✅ Why is Frequency Encoding used?
* Works well for high-cardinality categorical columns
* Faster than one-hot encoding
* Avoids creating too many dummy columns
* Useful for tree-based models (Random Forest, XGBoost, LightGBM)

```python
df = ds.copy()
```

```python
freq = df['Area'].value_counts()
freq
```

```python
df['Area'] = df['Area'].map(freq)
```

# target encoder used in the classification problem in the target column 

*  ✅ What is Target Encoding?
* Target Encoding replaces each category with the mean of the target variable for that category.
* It is mostly used for classification or regression when you have categorical columns.
