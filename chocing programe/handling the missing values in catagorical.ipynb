```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
ds = pd.read_csv('StudentPerformanceFactors.csv')
```

```python
ds['Teacher_Quality'].dtype
ds['Teacher_Quality'].isnull().sum()
ds['Teacher_Quality'].unique()
```

```python
ds['Teacher_Quality'].vlaue_counts().plot.bar()
ds['Teacher_Quality'].vlaue_counts().sort_values(ascending = True).plot.bar()
ds['Teacher_Quality'].vlaue_counts().plot.bar(kind = 'kde')
```

```python
dq =ds['Teacher_Quality'].mode()[0]
```

```python
ds['Teacher_Quality'] = ds['Teacher_Quality'].fillna(dq)
```

```python
ds['Teacher_Quality'].isnull().sum()
```

```python
ds['Teacher_Quality'].value_counts().plot(kind = 'kde')
```

```python
# plotting 
fig = plt.figure()
a = fig.add_subplot(111)
ds['Teacher_Quality'].value_counts().plot(kind = 'kde', ax = a)

ds['Teacher_Quality'].plot(kind = 'kde',ax = a, color = 'red')

plt.legend()
plt.show()
```
