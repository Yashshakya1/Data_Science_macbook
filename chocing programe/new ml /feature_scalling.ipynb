```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import MinMaxScaler,StandardScaler,RobustScaler
```

```python
df = pd.DataFrame({
    'age':[18,22,30,45,60,80],
    'salary':[20000,25000,40000,60000,120000,1000000]
})
print(df)
```

```python
mms = MinMaxScaler()
df_minmax = pd.DataFrame(mms.fit_transform(df) , columns=df.columns)
print(df_minmax)
```

# Standard scaler 

```python
standard_scaler = StandardScaler()
ss_result = pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)
print("\n standard scaled data :- ")
ss_result
```

# robust scaler 

```python
robust_scaler = RobustScaler()
rs_result= pd.DataFrame(robust_scaler.fit_transform(df),columns=df.columns)
rs_result
```
