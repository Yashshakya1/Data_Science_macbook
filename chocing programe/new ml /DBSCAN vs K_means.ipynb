```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd 
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN,KMeans
```

# dataset for non linear circle 

```python
X,_ = make_circles(n_samples=500, factor=0.5,noise=0.05,random_state=42)
```

# apply - kmeans 

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, init="k-means++", random_state=42)
kmeans_labels = kmeans.fit_predict(X)
```

```python
# Apply DBSCAN 
dbscan = DBSCAN(eps = 0.25, min_samples=5)
dbscan_labels = dbscan.fit_predict(X)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# kmeans plot 
axes[0].scatter(X[:,0], X[:,1], c=kmeans_labels, cmap="plasma", s=50, edgecolor="k")
axes[0].set_title("Kmeans (fails on circle)")

# dbscan plot
axes[1].scatter(X[:,0], X[:,1], c=dbscan_labels, cmap="plasma", s=50, edgecolor="k")
axes[1].set_title("DBSCAN")
```
