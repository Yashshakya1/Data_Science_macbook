```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.svm import SVC
```

```python
x= np.array([
    [2,2],
    [4,4],
    [2,0],
    [0,2],
]
)

y = np.array([1,1,-1,-1])
```

```python
print(x)
print(y)
```

```python
model= SVC(kernel='linear',C = 1e6)
model.fit(x,y)
```

```python
m = model.coef_[0]
c = model.intercept_[0]
support_vectors = model.support_vectors_
support_index = model.support_
```

```python
print(m)
print(c)
print(support_vectors)
print(support_index)
```

# calculate the margin 

```python
W_norm = np.linalg.norm(m)
distance_to_margin = 1.0/W_norm
full_margin = 2.0/W_norm
print("half margin",distance_to_margin)
print("full margin ",full_margin)
```
