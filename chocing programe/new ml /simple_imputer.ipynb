```python
import pandas as pd
import numpy as np

data = {
    'name': ['Ali', 'Sara', 'Ahmed', 'Fatima', 'Usman', 'Ayesha', 'Bilal', 'Zara'],
    'age': [25, np.nan, 35, 28, np.nan, 42, 30, np.nan],
    'salary': [50000, 60000, np.nan, 45000, 70000, np.nan, 55000, 48000],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR'],
    'city': ['Karachi', 'Lahore', np.nan, 'Karachi', 'Lahore', np.nan, 'Karachi', np.nan]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.impute import SimpleImputer
```

```python
mean_imputer = SimpleImputer(strategy='mean')
df_mean = df.copy()
df_mean[['age','salary']] = mean_imputer.fit_transform(df_mean[['age','salary']])
print("\n Imputed mean data:")
display(df_mean)
```

```python
median_imputer = SimpleImputer(strategy='median')
df_median = df.copy()
df_median[['age','salary']] = mean_imputer.fit_transform(df_median[['age','salary']])
print("\n Imputed median data:")
display(df_median)
```

```python
mode_imputer = SimpleImputer(strategy='most_frequent')
df_mode = df.copy()
df_mode[['city']] = mode_imputer.fit_transform(df_mode[['city']])
print("\n Imputed mode data:")
display(df_mode) # this is the jupyter notebook function which is used for showing the table 
```

```python
import random
df_random = df.copy()
def random_impute(series):
    filled = series.dropna().tolist()
    return series.apply(lambda x: random.choice(filled) if pd.isnull(x) else x)

for col in ['age','salary','city']:
    df_random[col] = random_impute(df_random[col])

print("\n random impute data :")
display(df_random)
```
