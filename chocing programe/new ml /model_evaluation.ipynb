* 👉 Precision = Pakde hue log Sahi nikle ya nahi?
* 👉 Recall = Jitne chor the, unme se kitne pakde?

| Metric        | Question                                                  |
| ------------- | --------------------------------------------------------- |
| **Precision** | “Jitno ko POSITIVE bola, unme se kitne sach me POSITIVE?” |
| **Recall**    | “Total POSITIVE me se kitne detect kiye?”                 |
