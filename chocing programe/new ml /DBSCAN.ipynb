```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.datasets import make_moons,make_circles,make_blobs
```

```python
x_moons,y_moons = make_moons(n_samples = 300,noise = 0.05,random_state=42)
```

```python
# make circle
# make circle
x_circles, y_circles = make_circles(
    n_samples = 300,
    factor = 0.5,
    noise = 0.05,
    random_state = 42
)
```

```python
# 3 make_blobs
x_blobs,y_blobs = make_blobs(n_samples = 300,centers = 3,cluster_std=0.6,random_state=42)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# plot moons
axes[0].scatter(x_moons[:,0], x_moons[:,1], c=y_moons, cmap="plasma", s=40, edgecolors="k")
axes[0].set_title("make_moons")

# plot circles
axes[1].scatter(x_circles[:,0], x_circles[:,1], c=y_circles, cmap="plasma", s=40, edgecolors="k")
axes[1].set_title("make_circles")

# plot blobs
axes[2].scatter(x_blobs[:,0], x_blobs[:,1], c=y_blobs, cmap="plasma", s=40, edgecolors="k")
axes[2].set_title("make_blobs")

plt.show()
```

# 🔑 DBSCAN depends on 2 parameters
# 1️⃣ eps (ε) – Radius
* How close two points must be to be considered neighbors.
#2️⃣ min_samples
* Minimum number of neighboring points required to form a dense region (cluster core).

```python
from sklearn.cluster import DBSCAN
```

```python
# step 1
X, _ = make_moons(n_samples=300,noise = 0.05,random_state=42)

# step 2: apply dbscan 
db = DBSCAN(eps=0.2,min_samples=5) 

db.fit(X)
```
