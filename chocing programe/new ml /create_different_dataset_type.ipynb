```python
import pandas as pd
import json
import sqlite3


data = {
    'id'  : [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie'],
    'age' : [25, 30, 35],
    'city': ['New York', 'London', 'Paris']
}
```

```python
ds = pd.DataFrame(data)
ds
```

```python
# save csv
ds.to_csv('data.csv', index=False)
# save json
ds.to_json('data.json', orient='records', indent=4)
ds.to_excel("data.xlsx",index = False)
conn = sqlite3.connect("data.db")
ds.to_sql("data",conn,if_exists="replace",index = False)
```
