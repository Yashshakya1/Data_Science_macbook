```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

ds = pd.DataFrame({
    'Size': ['Small', 'mediam', 'large', 'small'],
    'education': ['primary', 'secondary', 'graduate', 'postgraduate']
})

# Fix 1: categories list of lists mein honi chahiye
# Fix 2: case match karna zaroori hai — 'Small' != 'small'
oe_size = OrdinalEncoder(categories=[['small', 'mediam', 'large']])
ds['Size'] = ds['Size'].str.lower()  # Sab lowercase kar do
ds['new_size'] = oe_size.fit_transform(ds[['Size']])

# Education bhi encode karo
oe_edu = OrdinalEncoder(categories=[['primary', 'secondary', 'graduate', 'postgraduate']])
ds['new_education'] = oe_edu.fit_transform(ds[['education']])

print(ds)
```
