```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
ds = pd.read_csv('Kmeanssalry.csv')
ds
```

```python
ds.describe()
```

```python
plt.subplot(1,2,1)
sns.histplot(ds['Age'], kde = True, color = 'red')
plt.subplot(1,2,2)
sns.histplot(ds['Income'], kde = True, color = 'green')
plt.show()
```

```python
# don't compare because both coloumns are not normalization
```

```python
plt.figure(figsize=(8,5))

sns.kdeplot(ds['Age'], color='red', label='Age')
sns.kdeplot(ds['Income'], color='green', label='Income')

plt.title("Age vs Income KDE in One Plot")
plt.legend()
plt.show()
```

```python
sc = StandardScaler()
ds['income_scalling'] =  sc.fit_transform(ds[['Income']])
ds
```

```python
mms = MinMaxScaler()
ds['age_scalling'] = mms.fit_transform(ds[['Age']])
ds
```

```python
sns.kdeplot(ds['age_scalling'] , color = 'red')
sns.kdeplot(ds['income_scalling'] , color = 'green')
```
